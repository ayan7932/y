```python
flights = pd.read_csv("https://vincentarelbundock.github.io/Rdatasets/csv/nycflights13/flights.csv")
```

```python
print(flights.head())
```

```python
print(flights.shape)
```

```python
print(flights.info())
```

```python
print(flights.isnull().sum())
```

```python
#Q1
q1 = flights.dropna(subset=['dep_delay'])
q1 = q1.loc[q1['dep_delay'] >= 120]
q1.head()
```

```python
#Q2
#no clean needed if dest has non missing values
print(flights['dest'].isnull().sum())
```

```python
q2 = flights.loc[(flights['dest'] == 'IAH') | (flights['dest'] == 'HOU')]
```

```python
q2.head()
```

```python
#q3
q3 = flights.dropna(subset=['arr_delay', 'dep_delay'])
```

```python
q3 = q3.loc[(q3['arr_delay'] > 120) & (q3['dep_delay'] <= 0)]
```

```python
q3.head()
```

```python
#Q4
q4 = flights.dropna(subset=['dep_delay', 'arr_delay']).copy()
```

```python
q4['time_gained'] = q4['dep_delay'] - q4['arr_delay']
```

```python
q4 = q4.loc[(q4['dep_delay'] >= 60) & (q4['time_gained'] > 30)]
```

```python
q4.head()
```

```python
#Q5
# Q5 (1) Local clean
q5 = flights.dropna(subset=['dep_time'])

# Q5 (2) Filter for dep_time between 0 and 600
q5 = q5.loc[(q5['dep_time'] >= 0) & (q5['dep_time'] <= 600)]

# Q5 (3) Show head
q5.head()
```

```python
# Q6 (1) Local clean
q6 = flights.dropna(subset=['dep_delay'])

# Q6 (2) Sort by dep_delay descending
q6 = q6.sort_values(by='dep_delay', ascending=False)

# Q6 (3) Show head
q6.head()
```
